```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1=pd.read_csv("data.csv")
df1
```

```python
df1.info()
```

```python
df1.describe()
```

```python
df1.isnull().sum()
```

```python
df1=df1.dropna(subset=['Number of Doors'])
```

```python
df1['Engine HP']=df1['Engine HP'].fillna(df1['Engine HP'].mean())
```

```python
df1=df1.dropna(subset=['Engine Fuel Type'])
```

```python
df1['Engine Cylinders']=df1['Engine Cylinders'].fillna(df1['Engine Cylinders'].mean())
```

```python
df1.isnull().sum()
```

```python
df1.shape
```

```python
df1.columns
```

```python
df1=df1.drop(columns=['Market Category'],axis=1)
```

```python
df1.columns
```

```python
df1.isnull().sum()
```

```python
df1.shape
```

```python
df1.drop_duplicates(inplace=True)
```

```python
df1.shape
```

```python
df1['Make'].value_counts()
```

```python
plt.figure(figsize=(12,7))
sns.barplot(x=df1['Make'].value_counts().index,y=df1['Make'].value_counts(),color='purple')
plt.title("Number of Cars by Make")
plt.xlabel("Car's Name")
plt.xticks(rotation=90)
plt.show()
```

```python
df=df1.groupby('Make').agg({'MSRP':'mean'}).sort_values(by='MSRP',ascending=False)
df
```

```python
df2=df1.groupby(['Vehicle Style'])['Popularity'].agg('mean').sort_values(ascending=False)
df2
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=df2.index,y=df2,color='red')
plt.title("Popularity of Vehicle Style")
plt.xlabel("Vehicle Style")
plt.xticks(rotation=90)
plt.show()
```

```python
df3=df1.groupby('Make')['Popularity'].agg('mean').sort_values(ascending=False).reset_index()
df3
df3
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=df3['Make'],y=df3['Popularity'],color='green')
plt.title("Popularity of Make")
plt.xlabel("Make")
plt.xticks(rotation=90)
plt.show()
```
